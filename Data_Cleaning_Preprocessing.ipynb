```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## 2. Create the Sample Dataset

In this step, we create a sample employee dataset using Pandas. The dataset contains information about employee names, ages, cities, salaries, and departments.

We intentionally include missing values, duplicate records, and inconsistent text formatting to demonstrate common data-cleaning techniques.


```python
# Create a sample dataset with data-quality issues

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Bob', 'Frank', 'Grace', 'Henry', 'Ivy'],
    'Age': [25, 30, np.nan, 28, 35, 30, 40, 27, np.nan, 29],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai',
             'Mumbai', 'Delhi', 'Pune', 'Chennai', 'Delhi'],
    'Salary': [30000, 45000, 50000, np.nan, 60000,
               45000, 70000, 35000, 55000, 40000],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR',
                   'HR', 'IT', 'Finance', 'IT', 'HR']
}

df = pd.DataFrame(data)

# Introduce inconsistent formats
df.loc[6, 'City'] = ' delhi '
df.loc[7, 'Department'] = 'finance'

# Display the dataset
print("Original Dataset:")
display(df)
```

## 3. Inspect the Dataset

Before cleaning the data, we examine its structure, dimensions, column names, and data types. We also check for missing values and duplicate records.

This initial inspection helps identify data-quality issues that need to be addressed before further analysis.


```python
# Check the dataset dimensions
print("Dataset shape:", df.shape)

# Check column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
df.info()

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

## 4. Handle Missing Values

Missing values can affect the accuracy and reliability of data analysis. In this step, we replace missing values in the `Age` and `Salary` columns with their respective median values.

The median is used because it is less sensitive to extreme values than the mean. After replacement, we check the dataset to ensure that no missing values remain.


```python
# Handle missing values using the median

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Verify that missing values have been handled
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDataset after handling missing values:")
display(df)
```

## 5. Remove Duplicate Records

Duplicate records can lead to inaccurate analysis and misleading results. In this step, we identify and remove duplicate rows using Pandas' `drop_duplicates()` method.

After removing duplicates, we check the dataset again to confirm that no duplicate rows remain.


```python
# Count duplicate rows before removal
print("Duplicate rows before cleaning:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Verify duplicate removal
print("Duplicate rows after cleaning:", df.duplicated().sum())

print("\nDataset shape after removing duplicates:", df.shape)

display(df)
```

## 6. Standardize Text Values

Inconsistent text formatting can cause the same value to be treated as different categories. For example, `delhi`, `Delhi`, and `delhi` may be interpreted as different values.

In this step, we use `strip()` to remove extra spaces and `title()` to standardize capitalization in the `City` and `Department` columns.


```python
# Standardize text formatting in City and Department columns

df['City'] = df['City'].str.strip().str.title()
df['Department'] = df['Department'].str.strip().str.title()

# Display the corrected values
print("Unique cities:", df['City'].unique())
print("Unique departments:", df['Department'].unique())

print("\nDataset after standardizing text:")
display(df)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nFinal Cleaned Dataset:")
display(df)
```

```python
df.to_csv('Cleaned_Dataset.csv', index=False)

print("Cleaned dataset saved successfully!")
```
